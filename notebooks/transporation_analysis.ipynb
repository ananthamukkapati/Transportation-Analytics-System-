## Total passengers

```python
total_passengers = df["Passengers"].sum()

print(
    "Total passengers:",
    total_passengers
)
```

```python
average_passengers = df["Passengers"].mean()

print(
    "Average passengers per trip:",
    round(average_passengers, 2)
)
```

## Analyze Revenue

```python
total_revenue = df["Revenue"].sum()

print(
    "Total revenue:",
    total_revenue
)
```

```python
average_revenue = df["Revenue"].mean()

print(
    "Average revenue per trip:",
    round(average_revenue, 2)
)
```

## Analyze Profit

```python
total_profit = df["Profit"].sum()

print(
    "Total profit:",
    total_profit
)
```

```python
average_profit = df["Profit"].mean()

print(
    "Average profit per trip:",
    round(average_profit, 2)
)
```

## Vehicle utilization

```python
# Average 
average_utilization = (
    df["Utilization_Percentage"].mean()
)

print(
    "Average vehicle utilization:",
    round(average_utilization, 2),
    "%"
)
```

```python
vehicle_utilization = (
    df.groupby("Vehicle_Type")[
        "Utilization_Percentage"
    ]
    .mean()
    .sort_values(ascending=False)
)

print(vehicle_utilization)
```

## Route performance

```python

route_analysis = df.groupby("Route").agg({
    "Trip_ID": "count",
    "Passengers": "sum",
    "Revenue": "sum",
    "Profit": "sum",
    "Delay_min": "mean",
    "Utilization_Percentage": "mean"
})
route_analysis
```

```python
# rename
route_analysis.columns = [
    "Total_Trips",
    "Total_Passengers",
    "Total_Revenue",
    "Total_Profit",
    "Average_Delay",
    "Average_Utilization"
]
route_analysis
```

## Revenue by route

```python
plt.figure(figsize=(12, 6))

route_revenue = (
    df.groupby("Route")["Revenue"]
    .sum()
    .sort_values(ascending=False)
)

route_revenue.plot(kind="bar")

plt.title("Revenue by Route")
plt.xlabel("Route")
plt.ylabel("Revenue")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()
```

## Passenger demand by route

```python
plt.figure(figsize=(12, 6))

route_passengers = (
    df.groupby("Route")["Passengers"]
    .sum()
    .sort_values(ascending=False)
)

route_passengers.plot(kind="bar")

plt.title("Passenger Demand by Route")
plt.xlabel("Route")
plt.ylabel("Total Passengers")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()
```

## peak hour analysis

```python
hourly_demand = (
    df.groupby("Hour")["Passengers"]
    .sum()
)
hourly_demand
```

```python
plt.figure(figsize=(12, 6))

hourly_demand.plot(
    kind="line",
    marker="o"
)

plt.title("Passenger Demand by Hour")
plt.xlabel("Hour of Day")
plt.ylabel("Passengers")

plt.grid(True)

plt.tight_layout()

plt.show()
```

## Demand by day

```python
daily_demand = (
    df.groupby("Day_Name")["Passengers"]
    .sum()
)
day_order = [
    "Monday",
    "Tuesday",
    "Wednesday",
    "Thursday",
    "Friday",
    "Saturday",
    "Sunday"
]

daily_demand = daily_demand.reindex(day_order)

daily_demand
```

```python
plt.figure(figsize=(10, 5))

daily_demand.plot(
    kind="bar"
)

plt.title("Passenger Demand by Day")
plt.xlabel("Day")
plt.ylabel("Passengers")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()
```

## Delay Analysis

```python
average_delay = df["Delay_min"].mean()

print(
    "Average delay:",
    round(average_delay, 2),
    "minutes"
)
```

```python
route_delay = (
    df.groupby("Route")["Delay_min"]
    .mean()
    .sort_values(ascending=False)
)

route_delay
```

```python
plt.figure(figsize=(12, 6))

route_delay.plot(
    kind="bar"
)

plt.title("Average Delay by Route")
plt.xlabel("Route")
plt.ylabel("Average Delay (minutes)")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()
```

## Trip status analysis

```python
df["Status"].value_counts()
```

## Customer analysis

```python
customer_demand = (
    df.groupby("Customer_Type")["Passengers"]
    .sum()
    .sort_values(ascending=False)
)

customer_demand
```

```python
customer_revenue = (
    df.groupby("Customer_Type")["Revenue"]
    .sum()
    .sort_values(ascending=False)
)

customer_revenue
```

```python
plt.figure(figsize=(10, 6))

customer_revenue.plot(
    kind="bar"
)

plt.title("Revenue by Customer Type")
plt.xlabel("Customer Type")
plt.ylabel("Revenue")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()
```

## Correlation Analysis

```python
numeric_columns = [
    "Passengers",
    "Vehicle_Capacity",
    "Distance_km",
    "Duration_min",
    "Delay_min",
    "Fare",
    "Fuel_Cost",
    "Maintenance_Cost",
    "Revenue",
    "Total_Cost",
    "Profit",
    "Utilization_Percentage"
]
correlation = df[numeric_columns].corr()

correlation
```

```python
plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f"
)

plt.title("Transportation Analytics Correlation Matrix")

plt.tight_layout()

plt.show()
```
